```python
# 1. Create a Doc object from the file owlcreek.txt
import spacy

nlp = spacy.load("en_core_web_sm")
text = open("/content/owlcreek.txt").read()
doc = nlp(text)
```

```python
# 2. How many tokens are contained in the file ?
print("Number of tokens:", len(doc))
```

```python
#3. How many sentences are contained in the file?
sentences = list(doc.sents)
print("Number of sentences:", len(sentences))
```

```python
#4. Print the second sentence in the document?
print("Second sentence:", sentences[1].text)
```

```python
#5.For each token in the sentence above, print its text, POS tag, dep tag and lemma.
sentences = list(doc.sents)
second_sentence = sentences[1]

for token in second_sentence:
    print(token.text, token.pos_, token.dep_, token.lemma_)
```

```python
 #6.Write a matcher called 'Swimming' that finds both occurrences of the phrase "swimming vigorously" in the text.
import spacy
from spacy.matcher import Matcher

nlp = spacy.load("en_core_web_sm")

text = open("/content/owlcreek.txt").read()
doc = nlp(text)


matcher = Matcher(nlp.vocab)
pattern = [{"LOWER": "swimming"}, {"LOWER": "vigorously"}]
matcher.add("Swimming", [pattern])

matches = matcher(doc)

if matches:
    for match_id, start, end in matches:
        span = doc[start:end]
        print(f"Match found: '{span.text}' at position {start}-{end}")
else:
    print("No matches found ")
```

```python
#7. the text surrounding each found match.
# Apply the matcher to the doc
matches = matcher(doc)

# Print sentence context around each match
for match_id, start, end in matches:
    span = doc[start:end]
    sentence = span.sent.text
    print(f"Match: '{span.text}'")
    print(f"Sentence: {sentence}\n")
```
